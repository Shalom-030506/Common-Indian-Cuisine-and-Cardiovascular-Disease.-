```python
df2 = df.filter(items = ["Dish Name","Calories (kcal)" ,"Protein (g)","Fats (g)"])
               
df2
```

```python
df2.sort_values(by = ["Calories (kcal)" ,"Protein (g)","Fats (g)"], ascending = False)
df2
```

### Foods that are high in Calories (Threshold limit per day for Calories is 2000 to 2800 per day)

```python
df3 = df[df["Calories (kcal)"]> 700].sort_values(by = ["Calories (kcal)"])
df3
```

### Food which carry high amount of Fats in them (Threshhold for Fats per day is  44–78g )

```python
df4 = df[df["Fats (g)"]>80].sort_values(by = ["Fats (g)"])
df4
```

### High Protein Food (Required protein is at 0.83 g per kg a day)

```python
df5 = df[df["Protein (g)"]>0.83].sort_values(by = ["Protein (g)"])
df5

```

```python
### The Bar graph shows food with high amount of calories 
```

```python
fig = px.bar(df3, x='Dish Name', y ='Calories (kcal)', color = 'Calories (kcal)',title = "Food that Contain High amount of calories ", template = "none",height = 500, width = 900)
fig.show()

```

```python
fig = px.line(df4, x="Dish Name", y="Fats (g)", title=' Food with high amount of fats ', template = "none",height = 500, width = 700)
fig.show()
```
